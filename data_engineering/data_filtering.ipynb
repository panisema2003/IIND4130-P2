```python
import pandas as pd
```

```python
cleaned_data = pd.read_csv("../data/clean_icfes_data_cesar.csv")
df = cleaned_data.copy()
df
```

```python
filtered_df = df[["periodo", "cole_area_ubicacion", "cole_bilingue", "cole_calendario", 
                  "cole_caracter", "cole_cod_dane_establecimiento", "cole_cod_dane_sede", # TODO: see if we remove cod Dane establecimiento and sede, because they are not useful for the model, but we can use them for EDA
                  "cole_cod_mcpio_ubicacion", "cole_jornada", "cole_naturaleza", # cole_genero removed because all are MIXTO
                  "estu_cod_reside_depto", "estu_cod_reside_mcpio", "estu_estadoinvestigacion", 
                  "estu_fechanacimiento", "estu_genero", "estu_nacionalidad", # TODO: Ver si dejamos nacionalidad, pues hay poquitos 
                  "fami_cuartoshogar", "fami_educacionmadre", "fami_educacionpadre", 
                  "fami_estratovivienda", "fami_personashogar", "fami_tieneautomovil", 
                  "fami_tienecomputador", "fami_tieneinternet", "fami_tienelavadora", 
                  "desemp_ingles", "punt_ingles", "punt_matematicas", 
                  "punt_sociales_ciudadanas", "punt_c_naturales", "punt_lectura_critica", 
                  "punt_global"
                  ]]
filtered_df
```

```python
# Set column types to clean nulls

filtered_df["estu_fechanacimiento"] = pd.to_datetime(
    filtered_df["estu_fechanacimiento"],
    errors="coerce",
    dayfirst=True
)

# Set column types to clean nulls
filtered_df = filtered_df.astype({
    "periodo": "category",

    "cole_area_ubicacion": "category", # boolean in the future
    "cole_bilingue": "category", # boolean in the future
    "cole_calendario": "category", # boolean in the future
    "cole_caracter": "category",

    "cole_cod_dane_establecimiento": "string",
    "cole_cod_dane_sede": "string",
    "cole_cod_mcpio_ubicacion": "string",

    "cole_jornada": "category",
    "cole_naturaleza": "category", # boolean in the future

    "estu_cod_reside_depto": "string", # keep only 20
    "estu_cod_reside_mcpio": "string",

    "estu_estadoinvestigacion": "category", # keep only PUBLICAR

    # estu_fechanacimiento handled above with pd.to_datetime(errors="coerce")

    "estu_genero": "category", # boolean in the future
    "estu_nacionalidad": "category",

    "fami_cuartoshogar": "string", # clean and send to category later

    "fami_educacionmadre": "category", # TODO: see if we can scale this variable since it has order
    "fami_educacionpadre": "category", # TODO: see if we can scale this variable since it has order
    "fami_estratovivienda": "category", # TODO: see if we can scale this variable since it has order

    "fami_personashogar": "category", # clean and send to category/int later

    "fami_tieneautomovil": "category", # boolean in the future
    "fami_tienecomputador": "category", # boolean in the future
    "fami_tieneinternet": "category", # boolean in the future
    "fami_tienelavadora": "category", # boolean in the future

    "desemp_ingles": "category", # TODO: see if we can scale this variable since it has order

    # Nullable integers
    "punt_ingles": "Int64",
    "punt_matematicas": "Int64",
    "punt_sociales_ciudadanas": "Int64",
    "punt_c_naturales": "Int64",
    "punt_lectura_critica": "Int64",
    "punt_global": "Int64"
})
filtered_df
```

```python
# Clean all rows with any null values already present in the data
filtered_df = filtered_df.dropna(axis=0, how="any").reset_index(drop=True)
filtered_df
```

```python
# export cleaned data to csv
filtered_df.to_csv("../data/filtered_icfes_data_cesar.csv", index=False)
```
